```python
#Import the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

print("All libraries imported successfully!")
```

```python
#Create a sample dataset
data = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Kiran",
             "Divya", "Ravi", "Meena"],
    "Department": ["CSE", "DS", "CSE", "DS", "ECE",
                   "DS", "ECE", "CSE"],
    "Study_Hours": [2, 5, 3, 6, 2, 7, np.nan, 4],
    "Marks": [55, 85, 65, 90, np.nan, 95, 60, 75],
    "Attendance": [70, 90, 75, 95, 65, 98, 72, np.nan]
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)
```

```python
#Perform exploratory data analysis (EDA):
print("Dataset shape:", df.shape)

print("\nColumn information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nFirst five rows:")
display(df.head())
```

```python
#Handle missing data:
print("Missing values before cleaning:")
print(df.isnull().sum())

numeric_columns = ["Study_Hours", "Marks", "Attendance"]

df[numeric_columns] = df[numeric_columns].fillna(
    df[numeric_columns].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())

display(df)
```

```python
#Create a bar chart:
department_marks = df.groupby(
    "Department"
)["Marks"].mean()

plt.figure(figsize=(8, 5))
department_marks.plot(kind="bar")

plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
#Create a histogram:
plt.figure(figsize=(8, 5))

plt.hist(df["Marks"], bins=5, edgecolor="black")

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python
#Create a scatter plot:
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Study_Hours"],
    df["Marks"],
    s=80
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.tight_layout()
plt.show()
```

```python
#Perform correlation analysis:
correlation = df[
    ["Study_Hours", "Marks", "Attendance"]
].corr()

print("Correlation matrix:")
display(correlation)

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
df.to_csv("cleaned_student_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
