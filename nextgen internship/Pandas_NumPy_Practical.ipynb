```python

#Import Pandas and NumPy
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Libraries imported successfully!")
```

```python
#Create a sample dataset

data = {
    "Name": ["Arun", "Priya", "Rahul", "Sneha", "Kiran", "Divya"],
    "Department": ["CSE", "DS", "CSE", "DS", "ECE", "DS"],
    "Marks": [85, 92, 76, 88, 67, 95],
    "Attendance": [90, 95, 80, 85, 70, 98]
}

df = pd.DataFrame(data)

print("Student Dataset:")
display(df)
```

```python
#Load CSV and Excel files
# Create sample files
df.to_csv("students.csv", index=False)
df.to_excel("students.xlsx", index=False)

# Load the CSV file
csv_data = pd.read_csv("students.csv")
print("CSV file loaded:")
display(csv_data)

# Load the Excel file
excel_data = pd.read_excel("students.xlsx")
print("Excel file loaded:")
display(excel_data)
```

```python
#Perform filtering
high_scorers = df[df["Marks"] > 85]

print("Students scoring above 85:")
display(high_scorers)
```

```python
#Perform sorting
sorted_students = df.sort_values(
    by="Marks",
    ascending=False
)

print("Students sorted by marks:")
display(sorted_students)
```

```python
#Perform grouping
department_average = df.groupby(
    "Department"
)["Marks"].mean()

print("Average marks by department:")
display(department_average)
```

```python
#NumPy operations
marks = np.array([85, 92, 76, 88, 67, 95])

print("Marks:", marks)
print("Mean:", np.mean(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard deviation:", np.std(marks))

print("Marks above 85:", marks[marks > 85])
print("Marks sorted:", np.sort(marks))
```
